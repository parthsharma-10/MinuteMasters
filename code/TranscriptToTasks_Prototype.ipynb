This interactive notebook demonstrates the **Two-Stage Grounded Prompt Workflow** for converting raw, messy meeting transcripts into structured decisions and verified action items (with task, owner, deadline, and source quotation evidence).

```python
import json

# Two-stage Prompt Definitions
PROMPT_P1_SYSTEM = """You are the MinuteMasters Core Extraction Engine. Extract all finalized decisions and actionable items from the meeting transcript below.
Return a valid JSON object matching this schema:
{
  "decisions": [
    {"decision_text": "string", "evidence_quote": "string"}
  ],
  "action_items": [
    {"task": "string", "owner": "string or 'not stated'", "deadline": "string or 'not stated'", "evidence_quote": "string"}
  ]
}
"""

PROMPT_P2_SYSTEM = """You are the MinuteMasters Quality & Verification Auditor. Your objective is zero-hallucination verification.
You cross-examine candidate meeting minutes against the raw transcript. You reject any action item or decision that lacks direct transcript consensus or contains unconfirmed assumptions.
"""
print("MinuteMasters Core Pipeline loaded successfully.")
```

## 1. Test Input: Club Orientation Planning (Test 1)

```python
sample_transcript = """Asha: We should hold the club orientation on 20 Oct.
Ravi: Agreed. I will book the seminar hall by Friday.
Meera: I can design the poster, will share it Monday.
Asha: Okay, then Sana please email the freshers list after the poster is ready."""

print("--- RAW TRANSCRIPT INPUT ---")
print(sample_transcript)
```

## 2. Stage 1: Candidate Extraction (Prompt P1 Output)

```python
candidate_output = {
  "decisions": [
    {"decision_text": "Club orientation will be held on 20th October.", "evidence_quote": "Asha: We should hold the club orientation on 20 Oct. Ravi: Agreed."}
  ],
  "action_items": [
    {"task": "Book the seminar hall", "owner": "Ravi", "deadline": "Friday", "evidence_quote": "Ravi: Agreed. I will book the seminar hall by Friday."},
    {"task": "Design and share the orientation poster", "owner": "Meera", "deadline": "Monday", "evidence_quote": "Meera: I can design the poster, will share it Monday."},
    {"task": "Email the freshers list", "owner": "Sana", "deadline": "After the poster is ready", "evidence_quote": "Asha: Okay, then Sana please email the freshers list after the poster is ready."}
  ]
}
print(json.dumps(candidate_output, indent=2))
```

## 3. Stage 2: Adversarial Verification (Prompt P2 Output)

```python
verified_report = """=== FINAL VERIFIED AUDIT REPORT ===
Audit Summary: 1 decision verified, 3 action items verified, 0 hallucinations.

| # | Action Item | Owner | Deadline | Status |
|---|---|---|---|---|
| 1 | Book the seminar hall | Ravi | Friday | [Verified Grounded] |
| 2 | Design and share orientation poster | Meera | Monday | [Verified Grounded] |
| 3 | Email freshers list | Sana | After poster is ready | [Verified Contingent] |"""
print(verified_report)
```
